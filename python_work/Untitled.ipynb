```python
# Exercise 1: Print all even numbers from 1 to 20
for i in range(1, 21):
    if i % 2 == 0:
        print(i, end=" ")
print()

# Exercise 2: Print all odd numbers from 1 to 20
for i in range(1, 21):
    if i % 2 != 0:
        print(i, end=" ")
print()

# Exercise 3: Print multiples of 5 from 5 to 50
for i in range(5, 51, 5):
    print(i, end=" ")
print()

# Exercise 4: Find the sum of numbers from 1 to 10
total_sum = sum(range(1, 11))
print("Sum 1 to 10:", total_sum)

# Exercise 5: Find the sum of even numbers from 1 to 100
even_sum = sum(i for i in range(1, 101) if i % 2 == 0)
print("Sum of even 1 to 100:", even_sum)

# Exercise 6: Find the sum of odd numbers from 1 to 100
odd_sum = sum(i for i in range(1, 101) if i % 2 != 0)
print("Sum of odd 1 to 100:", odd_sum)

# Exercise 7: Print the multiplication table of 7
for i in range(1, 11):
    print(f"7 x {i} = {7 * i}")

# Exercise 8: Take a number from the user and print its multiplication table
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")

# Exercise 9: Print numbers from 1 to 100 that are divisible by both 3 and 5
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print(i, end=" ")
print()

# Exercise 10: Count even and odd numbers between 1 and 50
even_count = 0
odd_count = 0
for i in range(1, 51):
    if i % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
print(f"Evens: {even_count}, Odds: {odd_count}")

# Exercise 11: Print the elements of a list in reverse order
data = [22, 33, 12, 32, 45, 13]
print("Reversed list:", data[::-1])

# Exercise 12: Find the largest number in a list using for loop
data = [22, 55, 12, 89, 34, 67]
largest = data[0]
for num in data:
    if num > largest:
        largest = num
print("Largest number:", largest)

# Exercise 13: Find the smallest number in a list
data = [22, 55, 12, 89, 34, 67]
print("Smallest number:", min(data))

# Exercise 14: Count vowels in a string
text = "programming"
vowels = "aeiouAEIOU"
count = sum(1 for char in text if char in vowels)
print("Vowel count:", count)

# Exercise 15: Count the number of spaces in a sentence
text = "Python is easy to learn"
spaces = text.count(" ")
print("Number of spaces:", spaces)
```

```python
# 1. Check Student Result
students = {
    "Rahul": 75,
    "Amit": 32,
    "Priya": 88,
    "Neha": 39,
    "Rohit": 55
}
for name, marks in students.items():
    status = "Pass" if marks >= 40 else "Fail"
    print(f"{name}: {status}")

# 2. Student Performance Using Nested If
students = {
    "Rahul": 85,
    "Amit": 35,
    "Priya": 72,
    "Neha": 48
}
for name, marks in students.items():
    if marks >= 80:
        cat = "Excellent"
    elif marks >= 60:
        cat = "Good"
    elif marks >= 40:
        cat = "Average"
    else:
        cat = "Fail"
    print(f"{name}: {cat}")

# 5. Check Even/Odd Values
data = {"a": 10, "b": 15, "c": 25, "d": 30, "e": 40}
for k, v in data.items():
    if v % 2 == 0:
        label = "Large Even" if v > 20 else "Even"
    else:
        label = "Odd"
    print(f"{k}: {label}")

# 6. Shopping Cart Classification
products = {
    "Mobile": 15000,
    "Mouse": 500,
    "Laptop": 60000,
    "Keyboard": 1200
}
for prod, price in products.items():
    if price >= 5000:
        cat = "Premium Product"
    elif price >= 1000:
        cat = "Expensive Product"
    else:
        cat = "Affordable Product"
    print(f"{prod}: {cat}")

# 7. Student Eligibility
students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}
for name, info in students.items():
    if info["attendance"] < 75:
        status = "Not Eligible"
    else:
        status = "Eligible" if info["marks"] >= 40 else "Failed in Exam"
    print(f"{name}: {status}")

# 8. Bank Account Classification
accounts = {
    "Rahul": 25000,
    "Amit": 5000,
    "Priya": 0,
    "Neha": -2000
}
for name, bal in accounts.items():
    if bal >= 10000:
        cat = "Premium Account"
    elif bal > 0:
        cat = "Regular Account"
    elif bal == 0:
        cat = "No Balance"
    else:
        cat = "Overdraft"
    print(f"{name}: {cat}")

# 9. Nested Dictionary - Employee Experience
# (Example data provided based on prompt specifications)
employees = {
    "Rahul": {"salary": 60000, "experience": 6},
    "Amit": {"salary": 55000, "experience": 3},
    "Priya": {"salary": 40000, "experience": 4},
    "Neha": {"salary": 30000, "experience": 1}
}
for name, info in employees.items():
    if info["salary"] >= 50000:
        role = "Senior" if info["experience"] >= 5 else "Mid-Level"
    else:
        role = "Experienced Junior" if info["experience"] >= 3 else "Fresher/Junior"
    print(f"{name}: {role}")

# 10. Student Marks & Attendance Grading
students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}
for name, info in students.items():
    if info["attendance"] < 75:
        grade = "Not Eligible"
    else:
        m = info["marks"]
        if m >= 80:
            grade = "Distinction"
        elif m >= 60:
            grade = "First Division"
        elif m >= 40:
            grade = "Pass"
        else:
            grade = "Fail"
    print(f"{name} {grade}")
```
